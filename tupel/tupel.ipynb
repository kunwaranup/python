Python Tuple


A tuple in Python is a collection used to store multiple values in a single variable. A tuple is similar to a list, but the main difference is that a tuple cannot be changed after it is created.


Tuples are ordered and can contain duplicate values and different data types.

```python
fruits = ("Apple", "Banana", "Mango")

print(fruits)
```

```python
print(type(fruits))
```

Tuple Can Store Different Data Types

```python
student = ("Anup", 25, 85.5, True)

print(student)
```

Accessing Tuple Items

Like lists, tuple indexing starts from 0.

```python
fruits = ("Apple", "Banana", "Mango")

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

```python
print(fruits[-1])
```

Tuple Cannot Be Changed

This is the most important difference between a list and a tuple.

```python
#fruits = ("Apple", "Banana", "Mango")

#fruits[1]="orange"
#TypeError: 'tuple' object does not support item assignment
```

```python
fruits = ("Apple", "Banana", "Mango")

print(len(fruits))
```

Tuple Unpacking

You can assign tuple values to multiple variables.

```python
student = ("Anup", 25, "Python")

name, age, skill = student

print(name)
print(age)
print(skill)
```
