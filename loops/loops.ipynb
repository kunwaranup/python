loops

A loop in Python is used to execute the same block of code repeatedly.

Loops are useful when you want to perform an operation multiple times without writing the same code again and again.

Python mainly has two types of loops:

1.for loop

2.while loop

for Loop

A for loop is generally used when you want to iterate over a sequence or repeat something a known number of times.

```python
# for variable in sequence:
```

Loop Through a List

A for loop can be used to access each item in a list.

```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

range() Function

range() is commonly used with for loops.

range(stop)

```python
for i in range(5):
    print(i)
```

```python
for i in range(1,10):
  print(i)
```

```python
for i in range(2, 10, 2):
    print(i)
```

while Loop

A while loop repeatedly executes code as long as a condition is True.

```python
count = 1

while count <= 5:
    print(count)
    count =count+1
```

```python
password = ""

while password != "1234":
    password = input("Enter password: ")

print("Login successful!")
```

break

break is used to stop a loop immediately.

```python
for i in range (1, 20):
  if i== 10:
    break
  print(i)
```

```python
for i in range(1, 10):

    if i == 5:
        continue

    print(i)
```

```python
total=0
marks=[50,60,40,80]
for mark in marks:
  total=total+mark

average=total/len(marks)
print(f"average:{average}")
```

```python
n=int(input("enter the number:"))
for i in range (1,11):
  print(i ,"*", n ,"=", i*n)
  i=i+1
```

Explanation:
- i = 1 initializes the variable.
- i <= 5 checks the condition.
- i += 1 increases i by one after each iteration.
- When i becomes 6, the condition becomes false, and the loop stops.
Important: If you forget to update the variable, a while loop may run forever. This is called an infinite loop.

Difference Between for and while Loops

for loop

Used to iterate over a sequence or range.

Often used when the number of iterations is known.

Automatically moves to the next item.


while loop

Used while a condition remains true.

Often used when the stopping condition determines the number of iterations.

You may need to update variables manually.


The break Statement

The break statement immediately terminates a loop.

```python
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

The continue Statement

The continue statement skips the current iteration and moves to the next one.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
#Print even numbers from 2 to 20.
for i in range(2, 21, 2):
    print(i)
```

```python
#Print numbers from 5 down to 1 using a while loop.
i = 5

while i >= 1:
    print(i)
    i -= 1
```
