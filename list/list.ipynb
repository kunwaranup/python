###python lists

A list in Python is a collection used to store multiple values in a single variable.

Lists are one of the most commonly used data types in Python.
can store mixed data type

Python lists can contain strings, numbers, Boolean values, and even other lists.

```python
student = ["Anup", 25, 85.5, True]
```

Accessing List Items

Python uses index numbers to access items.

The index starts from 0.

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
```

Negative Index

You can also access items from the end using negative indexes.

```python
print(fruits[-1])
print(fruits[-2])
```

Changing List Items

Lists are mutable, which means you can change their values.

```python
fruits = ["Apple", "Banana", "Mango"]

fruits[1] = "Orange"

print(fruits)
```

Adding Items


append()

Adds an item to the end of the list.

```python
fruits = ["Apple", "Banana"]

fruits.append("Mango")

print(fruits)
```

```python
#insert()
fruits = ["Apple", "Mango"]

fruits.insert(1, "Banana")

print(fruits)
```

emoving Item

Removes a specific value.

```python
fruits = ["Apple", "Banana", "Mango"]

fruits.remove("Banana")

print(fruits)
```

Finding the Length of a List

Use the len() function.

```python
fruits = ["Apple", "Banana", "Mango"]

print(len(fruits))
```

Sorting a List

Use sort() to arrange items.

```python
numbers = [5, 2, 8, 1, 3]

numbers.sort()

print(numbers)
```

```python
#for desending order
numbers.sort(reverse=True)

print(numbers)
```

List Slicing

Slicing allows you to get a portion of a list.

```python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]

print(fruits[1:4])
```
