Python Functions

A function in Python is a block of reusable code that performs a specific task. Instead of writing the same code repeatedly, you can define a function once and call it whenever you need it.
1. Why do we use functions?

- Code reusability: Write code once and use it many times.
- Less repetition: Avoid writing the same code repeatedly.
- Easy debugging: Find and fix errors more easily.
- Better organization: Divide a large program into smaller parts.

def — keyword used to define a function.



function_name — the name of the function.



() — parentheses used for parameters.



: — marks the beginning of the function body.



Indented code — the statements executed by the function.

```python
def greet():
    print("Hello, World!")

greet()
```

Function with a default parameter

A default parameter has a value that is used when no argument is provided for it.

```python
def greet(name="Student"):
    print("Hello", name)

greet()
greet("Ram")
```

Function with multiple parameters

A function can accept multiple inputs.

```python
def student(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student("Ram", 22, "Python")
```

```python
def calculate_percentage(math, science, english):
    total = math + science + english
    percentage = (total / 300) * 100

    print("Total marks:", total)
    print("Percentage:", percentage, "%")

calculate_percentage(80, 75, 85)
```

```python
#Write a function to calculate the factorial of a positive integer.
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial(5))
```
