### dictionary

A dictionary in Python is a collection used to store data in key-value pairs.

Unlike a list, which uses indexes such as 0, 1, 2, a dictionary uses keys to access its values.

```python
dictionary_name = {
    "key1": "value1",
    "key2": "value2"
}
```

```python
student = {
    "name": "Anup",
    "age": 25,
    "course": "Python",
    "marks": 85
}

print(student)
```

```python
print(type(student))
```

Accessing Dictionary Values                                     

Use the key to access a value.

```python
student = {
    "name": "Anup",
    "age": 25,
    "marks": 85,
    "rollno": 5,
    "class": 10
}

print(student["name"])
print(student["age"])
print(student["class"])
```

You can also use get():

```python
print(student.get("marks"))
```

nested dictionary

```python
students = {
    "student1": {
        "name": "hari",
        "age": 25
    },
    "student2": {
        "name": "Ram",
        "age": 22
    }
}

print(students["student1"]["name"])
```

adding new item

```python
employe = {
    "emp1": {
        "name": "shom",
        "age": 30
       },
    "emp2": {
        "name": "lokesh",
        "age":  40
    }
}
employe["emp1"]["salary"]=25000
print(employe)


```

CHANGING THE VALUE           

Dictionaries are mutable, so you can change their values.

```python
sts={
    "name":"bini",
    "age":25
    }

sts["age"]=30
print(sts)
```

Removing the items

```python
worker={
    "name":"hari",
    "age": 35,
    "address":"ktm"

    }
worker.pop("age")
print(worker)
```

You can also use del:

```python
del worker["address"]
print(worker)
```

Dictionary with Different Data Types

A dictionary can contain different types of values.

```python
student = {
    "name": "Anup",
    "age": 25,
    "marks": 85.5,
    "passed": True,
    "skills": ["Python", "JavaScript"]
}

print(student)
```

BMI CALCULATOR

```python
weight = int(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = weight / (height ** 2)

print("Your BMI is:", round(bmi, 2))

if bmi < 18.5:

    print("BMI category: Underweight")

elif bmi < 25:
    print("BMI category: Healthy range")

elif bmi < 30:
    print("BMI category: Overweight")

else:
    print("BMI category: Obesity")
```
