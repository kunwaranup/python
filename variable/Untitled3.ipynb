## variable


A variable in Python is a name used to store a value. Python does not require you to declare the variable type; the type is automatically determined from the value assigned.

```python
name ="anup"
age = 25
height = 5.6
```

```python
print("Name:", name)
print("age:", age)
print("height:", height)
```

Variable names should start with a letter or _, cannot contain spaces, and are case-sensitive.
For example, age, Age, and AGE are three different variables.
