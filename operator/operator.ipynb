##operator

An operator is a symbol or keyword used to perform an operation on values or variables.

1.Arthematic operator

```python
a=10
b=3
```

```python
print("addition:", a+b)
print("sub:",a-b)
print("multiply",a*b)
print("div:",a/b)
print("floor division:",a//b)
print("reminder:",a%b)
print("power:",a**b)
```

###assignment operator
used to assign or update value

```python
x = 10       # assign
x += 5       # x = x + 5
x -= 2       # x = x - 2
x *= 3       # x = x * 3
x /= 2       # x = x / 2
```

```python
x=10
x+= 5
print(x)
```
